## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Search question

What performance archetypes exist across the content inventory when content items are grouped using structured content and search-performance signals?

Unit of analysis

The unit of analysis is one content item/page.

Each page will be represented using measurable structured features such as search performance, engagement, content characteristics, freshness, and other available metadata.

Decision

The decision I want to support is:

Which types of content appear in the inventory, and which types may deserve different levels or kinds of attention?

Instead of treating every page in the same way, the cluster profiles could help a content or SEO team understand where different patterns exist.

Output

The intended output is:

A cluster assignment for each content item.
A profile describing the typical characteristics of each cluster.
A human-readable description of each archetype.
A possible action or review priority for each archetype.

I will decide the names of the archetypes only after examining the actual cluster profiles.

Action

The clusters could be used to prioritize further investigation.

For example, one group may show strong observed performance, while another may show high visibility but weaker engagement. Depending on the actual results, different groups could be candidates for monitoring, improvement, or deeper manual review.

The cluster itself will not automatically determine the action.

Cost of a wrong recommendation

A wrong recommendation could lead to wasted effort or missed opportunities.

If a healthy group of pages is incorrectly treated as needing major changes, time and resources could be wasted. If a group with weaker observed performance is missed, a potentially useful improvement opportunity could be overlooked.

For this reason, I would use the clusters as a decision-support tool, not as an automatic decision-maker.

Why data or ML can help

Clustering is useful to investigate because the page archetypes are not known in advance.

A rule-based approach would require me to define categories and thresholds before seeing the data. An unsupervised method can instead look for groups of content items that have similar structured characteristics.

The goal is therefore not simply to produce clusters. The clusters should also be understandable, reasonably stable, and useful for supporting a practical decision.

## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Basic dataset facts
total_pages = len(df)

missing_word_count = df["word_count"].isna().sum()
missing_search_volume = df["search_volume"].isna().sum()

missing_word_pct = missing_word_count / total_pages * 100
missing_search_pct = missing_search_volume / total_pages * 100

print(f"Total content items: {total_pages:,}")
print(
    f"Rows missing word_count: "
    f"{missing_word_count:,} ({missing_word_pct:.1f}%)"
)
print(
    f"Rows missing search_volume: "
    f"{missing_search_volume:,} ({missing_search_pct:.1f}%)"
)
```
